```python
import pandas as pd
```

```python
patients = pd.DataFrame({
    "Patient_ID": [1, 2, 3, 4, 5, 6],
    "Name": ["Rahul", "Anita", "Suresh", "Pooja", "Vikas", "Meena"],
    "Doctor_ID": [101, 102, 101, 103, 104, 102],
    "Age": [25, 34, np.nan, 45, 29, np.nan]
})

doctors = pd.DataFrame({
    "Doctor_ID": [101, 102, 103, 104],
    "Doctor": ["Dr. Sharma", "Dr. Mehta", "Dr. Khan", "Dr. Singh"],
    "Department": ["Cardiology", "Neurology", "Orthopedic", "Dermatology"]
})

followup = pd.DataFrame({
    "Patient_ID": [7, 8],
    "Name": ["Arjun", "Kavita"],
    "Doctor_ID": [103, 104],
    "Age": [38, 31]
})
```

```python
pd.isnull(patients).sum()
```

```python
m = patients["Age"].mean()
```

```python
patients["Age"] = patients["Age"].fillna(m)
```

```python
all_p = pd.concat([patients, followup])
```

```python
all_p
```

```python
all_patients = all_p.reset_index(drop=True)
```

```python
all_patients
```

```python
finaldf = pd.merge(all_p, doctors)
```

```python
data = {
    'Date': pd.date_range('2026-09-03', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values = "Sales",index = 'Region',columns="Product",aggfunc = 'median')
```

```python
pd.pivot_table(df,values = "Sales",index = 'Region',columns="Product",aggfunc = 'sum')
```

```python
df.groupby(["Region", "Product"])["Sales"].sum()
```

```python
pivot2 = pd.pivot_table(
    df,
    values=["Sales", "Units"],
    index="Region",
    columns="Product"
)
```

```python
pivot2
```

```python
pivot3 = pd.pivot_table(
    df,
    values=["Sales", "Units"],
    index="Region",
    columns="Product",
    aggfunc={
        "Sales": "sum",
        "Units": "sum"
    }
)

pivot3
```

```python
pivot4 = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc=["sum", "mean", "max"]
)
pivot4 
```

```python
pivot5 = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

pivot5
```

```python
pivot6 = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

pivot6
```

```python
pd.crosstab(df['Region'],df['Product'])
```
